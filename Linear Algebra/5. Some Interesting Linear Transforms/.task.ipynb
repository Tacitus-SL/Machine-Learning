# Inversion Matrix

Write code that generates a matrix that reverses the order of coordinates of a vector in $N$-dimensional space:

$$
\begin{bmatrix}
x_1 \\
\vdots \\
x_N
\end{bmatrix}
\rightarrow
A
\rightarrow
\begin{bmatrix}
x_N \\
\vdots \\
x_1
\end{bmatrix}
$$

```python
# Edit this cell in task.ipynb.
def inversion_matrix(N):
    res = np.zeros([N, N])
    ## YOUR CODE HERE
    # -- start --
    res = np.eye(N)
    res = res[:, ::-1]
    # -- end --

    return res
```

```python
# Let us check, how does it work

x = np.arange(5)
print('x:', x)
print('Ax:', inversion_matrix(5) @ x)
```

# Rolling Sum Matrix

Write code that generates a matrix that performs a rolling sum of the coordinates of a vector in $N$-dimensional space:

$$
\begin{bmatrix}
x_1 \\
x_2 \\
\vdots \\
x_N
\end{bmatrix}
\rightarrow
A
\rightarrow
\begin{bmatrix}
x_1 \\
x_1 + x_2 \\
\vdots \\
x_1 + x_2 + \dots + x_N
\end{bmatrix}
$$

```python
# Edit this cell in task.ipynb.
def rolling_sum_matrix(N):
    res = np.zeros([N, N])
    ## YOUR CODE HERE
    # -- start --
    res = np.tril(np.ones([N, N]))
    # -- end --

    return res
```

```python
# Let us check, how does it work

x = np.arange(5)
print('x:', x)
print('Ax:', rolling_sum_matrix(5) @ x)
```

# Neighbors Difference Matrix

Write code that generates a matrix that computes the differences between adjacent elements of a vector in $N$-dimensional space:

$$
\begin{bmatrix}
x_1 \\
\vdots \\
x_N
\end{bmatrix}
\rightarrow
A
\rightarrow
\begin{bmatrix}
x_2 - x_1 \\
\vdots \\
x_N - x_{N-1}
\end{bmatrix}
$$

```python
# Edit this cell in task.ipynb.
def neighbors_difference_matrix(N):
    res = np.zeros([N, N])
    ## YOUR CODE HERE
    # -- start --
    res = np.diag([1] * (N - 1), k=1) - np.eye(N, N)
    res = res[:N-1, :]
    # -- end --
    return res
```

```python
# Let us check, how does it work

x = np.arange(5)
print('x:', x)
print('Ax:', neighbors_difference_matrix(5) @ x)
```

# Permutation Matrix

You are given a permutation $\sigma$, which specifies the order in which the elements should appear.

Write a program that generates a matrix that permutes
the coordinates of a vector:

$$
\begin{bmatrix}
x_1 \\
x_2 \\
\vdots \\
x_N
\end{bmatrix}
\rightarrow
A
\rightarrow
\begin{bmatrix}
x_{\sigma_1} \\
x_{\sigma_2} \\
\vdots \\
x_{\sigma_N}
\end{bmatrix}
$$

For example, if you are given a permutation $\sigma = (3, 4, 2, 1)$,
the resulting vector should have the following form:

$$[x_3, x_4, x_2, x_1]^T$$

```python
# Edit this cell in task.ipynb.
def permutation_matrix(order):
    N = len(order)
    res = np.zeros([N, N])
    ## YOUR CODE HERE
    # -- start --
    res[np.arange(len(order)), order] = 1
    # -- end --

    return res
```

```python
# Let us check, how does it work

order = np.random.permutation(5)
x = np.arange(5)
print('x:', x)
print('order:', order)
print('Ax:', permutation_matrix(order) @ x)
```

```python
answer['task4'] = []
for one_input in inputs['task4']:
    answer['task4'].append(permutation_matrix(**one_input))
```

# 2D Rotation Matrix

Write code that, given an angle $\theta$, generates a matrix that rotates any 2D vector counterclockwise by $\theta$ without changing its length.

```python
# Edit this cell in task.ipynb.
def rotation_2D_matrix(phi):
    res = np.zeros([2, 2])
    ## YOUR CODE HERE
    # -- start --
    res[0, 0] = np.cos(phi)
    res[0, 1] = - np.sin(phi)
    res[1, 0] = np.sin(phi)
    res[1, 1] = np.cos(phi)
    # -- end --

    return res
```

```python
# Let us check, how does it work

order = np.random.permutation(5)
x = np.array([1, 1])

print('x:', x)
print('order:', order)
print('Ax:', rotation_2D_matrix(np.pi / 4) @ x)
```
